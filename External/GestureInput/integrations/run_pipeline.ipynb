```python
import subprocess
import sys
from pathlib import Path
HOST = '127.0.0.1'
PORT = 5555

DEVICE = 0
WIDTH = 960
HEIGHT = 540

NUM_HANDS = 2

MODEL_PATH = 'model/hand_landmarker.task'
CLASSIFIER_PATH = 'model/keypoint_classifier/keypoint_classifier.tflite'
LABELS_PATH = 'model/keypoint_classifier/keypoint_classifier_label.csv'

MIN_DETECTION_CONFIDENCE = 0.7
MIN_TRACKING_CONFIDENCE = 0.5

HEIGHT_DIFF_THRESHOLD = 0
MIN_GESTURE_CONFIDENCE = 0.6

# Strefa spoczynku (półokrąg u dołu ekranu): obie ręce w strefie => inactive
REST_ZONE_RADIUS = 300    # promień r [px], 0 = wyłączona
REST_ZONE_Y_OFFSET = 120   # środek okręgu tyle px PONIŻEJ dolnej krawędzi ekranu
GESTURE_COOLDOWN = 0.4

# EEG (folder eeg/, opis w eeg/README.md)
EEG = 'headset'            # 'headset' (BCI Core-8), 'sim' (losowy sygnał), 'off' (sama kamera)
EEG_FALLBACK = 'block'     # EEG niedostępne: 'block' = brak umiejętności, 'cv' = sama kamera
EEG_MIN_WINDOWS = 6        # Heal/Choke/Lightning: okna EEG (co 0.25 s) przed startem
EEG_MIN_WINDOWS_PUSH = 3   # Push: okna EEG przed wyzwoleniem
EEG_THRESHOLD_OFFSET = 0.0 # + = mniej fałszywych aktywacji
EEG_RELEASE = 1.5          # s utraty wyobrażenia, po których trwająca umiejętność gaśnie (0 = tylko gest)
CV_STABLE = 0.15           # s stabilnego gestu z kamery, zanim się liczy
CV_GRACE = 0.35            # krótsze przerwy w rozpoznaniu gestu są ignorowane (antymigotanie)
CV_REJOIN = 1.0            # ten sam gest wraca w ciągu tylu s = ciąg dalszy (EEG się sumuje)
DECISION = 'gate'          # 'gate' = czekaj na okna EEG z czasu gestu; 'instant' = porównaj wynik EEG ruchu z progiem w chwili wykrycia (powyżej -> TCP, poniżej -> ignoruj)
EEG_CHECK_WINDOWS = 4      # 'instant': średnia z tylu ostatnich okien EEG
LEGACY_EVENTS = False      # True = stary protokół gesture_started/gesture_ended bez EEG

RECONNECT_DELAY = 2.0

NO_DISPLAY = False
NO_FLIP = False

MAX_FRAMES = 0

DRY_RUN = False
VERBOSE = False

ARGS = [
    '--host', str(HOST),
    '--port', str(PORT),
    '--reconnect_delay', str(RECONNECT_DELAY),

    '--device', str(DEVICE),
    '--width', str(WIDTH),
    '--height', str(HEIGHT),

    '--model_path', MODEL_PATH,
    '--classifier_path', CLASSIFIER_PATH,
    '--labels_path', LABELS_PATH,

    '--num_hands', str(NUM_HANDS),

    '--min_detection_confidence', str(MIN_DETECTION_CONFIDENCE),
    '--min_tracking_confidence', str(MIN_TRACKING_CONFIDENCE),

    '--height_diff_threshold', str(HEIGHT_DIFF_THRESHOLD),
    '--min_gesture_confidence', str(MIN_GESTURE_CONFIDENCE),

    '--rest_zone_radius', str(REST_ZONE_RADIUS),
    '--rest_zone_y_offset', str(REST_ZONE_Y_OFFSET),
    '--gesture_cooldown', str(GESTURE_COOLDOWN),

    '--eeg', EEG,
    '--eeg_fallback', EEG_FALLBACK,
    '--eeg_min_windows', str(EEG_MIN_WINDOWS),
    '--eeg_min_windows_push', str(EEG_MIN_WINDOWS_PUSH),
    '--eeg_threshold_offset', str(EEG_THRESHOLD_OFFSET),
    '--eeg_release', str(EEG_RELEASE),
    '--cv_stable', str(CV_STABLE),
    '--cv_grace', str(CV_GRACE),
    '--cv_rejoin', str(CV_REJOIN),
    '--decision', DECISION,
    '--eeg_check_windows', str(EEG_CHECK_WINDOWS),

    '--max_frames', str(MAX_FRAMES),
]

if LEGACY_EVENTS:
    ARGS.append('--legacy_events')

if NO_DISPLAY:
    ARGS.append('--no_display')

if NO_FLIP:
    ARGS.append('--no_flip')

if DRY_RUN:
    ARGS.append('--dry_run')

if VERBOSE:
    ARGS.append('--verbose')
```

```python
NOTEBOOK_DIR = Path.cwd()
ROOT = NOTEBOOK_DIR if (NOTEBOOK_DIR / 'integrations').is_dir() else NOTEBOOK_DIR.parent
SCRIPT = ROOT / 'integrations' / 'gesture_pipeline.py'

proc = subprocess.run(
    [sys.executable, str(SCRIPT), *ARGS],
    cwd=str(ROOT),
)

print('Exit code:', proc.returncode)
```
